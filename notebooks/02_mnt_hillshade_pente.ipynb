# Expérience 2 — Détection à partir du MNT, du hillshade 315° et de la pente

Dans cette expérience, j'associe l'altitude brute du **MNT** à deux informations dérivées du relief : le **hillshade calculé avec un azimut de 315°** et la **pente**. L'objectif est d'étudier si ces représentations morphologiques apportent au U-Net des indices supplémentaires pour reconnaître les linéaments.

Ce notebook permet deux utilisations :

- **entraîner un nouveau modèle** avec vos propres données et un masque de linéaments ;
- **appliquer un modèle déjà entraîné** à une nouvelle zone pour produire des prédictions.

### Données nécessaires

| Utilisation | Couches d'entrée | Masque de référence | Modèle enregistré |
|---|---:|---:|---:|
| Nouvel entraînement | requises | requis | non |
| Prédiction sur une nouvelle zone | requises | facultatif | requis |

Le modèle attend trois GeoTIFF à une bande, dans l'ordre MNT, hillshade 315° et pente. La pente et le hillshade doivent déjà être calculés. Le notebook les contrôle et les utilise, mais ne les produit pas à partir du MNT. Le masque sert de vérité de référence et doit être aligné avec toutes les entrées.

### Parcours du notebook

Le traitement suit l'ordre suivant : contrôle des données, visualisation, découpage en patches, répartition train/validation/test, normalisation, augmentation, construction du U-Net, entraînement, évaluation et sauvegarde. Exécutez les cellules dans l'ordre. Après une modification des données ou des paramètres, reprenez à partir de l'étape 2.

### C. Comprendre les paramètres principaux

| Paramètre | Signification | Conseil d'utilisation |
|---|---|---|
| `TAILLE_PATCH` | Hauteur et largeur, en pixels, de chaque image donnée au U-Net. | `64` reproduit le découpage de l'expérience. Utilisez un multiple de 16, au minimum 32. |

| `PERCENTILES` | Bornes robustes utilisées pour normaliser chaque canal entre 0 et 1. | `(2, 98)` limite l'influence des valeurs extrêmes. |

| `EPOQUES` | Nombre de passages complets sur les données d'entraînement. | `2` permet un contrôle court ; `150` correspond au réglage de l'expérience. |
| `BATCH_SIZE` | Nombre d'images traitées simultanément. | Réduisez la valeur si la mémoire est insuffisante. |
| `LEARNING_RATE` | Amplitude des mises à jour des poids. | La valeur utilisée ici est `0,0001`. |
| `WEIGHT_DECAY` | Régularisation appliquée par AdamW. | Conservez `0,0001` pour reproduire le réglage proposé. |
| `DEVICE` | Matériel utilisé pour les calculs. | `auto` choisit CUDA si disponible, sinon le CPU. |

### D. Activer l'analyse vectorielle, si souhaité

Cette analyse est indépendante de l'entraînement. Pour l'utiliser, placez **au choix** un shapefile complet (`lineaments.shp`, `.shx`, `.dbf`, `.prj`) ou un GeoPackage unique (`lineaments.gpkg`) dans `data/ma_zone`, puis choisissez `ACTIVER_ANALYSE_LINEAMENTS=True` et adaptez `CHEMIN_LINEAMENTS`. Avec un GeoPackage à plusieurs couches, renseignez aussi `COUCHE_LINEAMENTS`. Si vous voulez uniquement entraîner le modèle, laissez l'option à `False`.


```python
CONFIGURATION = RACINE / "configs/mnt_hillshade315_pente.json"
MODE = "entrainer"  # "entrainer" ou "modele_enregistre"
UTILISER_DRIVE = False  # True dans Colab si vos fichiers sont sur Google Drive.
DOSSIER_DONNEES = RACINE / "data/ma_zone"
DOSSIER_SORTIE = RACINE / "resultats/mnt_hillshade315_pente"
ENTREES = {
    "mnt": DOSSIER_DONNEES / "mnt.tif",
    "hillshade": DOSSIER_DONNEES / "hillshade_315.tif",
    "pente": DOSSIER_DONNEES / "pente.tif",
}
MASQUE = DOSSIER_DONNEES / "masque.tif"  # None permis uniquement pour prédire sans évaluer.

# Analyse vectorielle facultative des linéaments de référence.
ACTIVER_ANALYSE_LINEAMENTS = False  # True uniquement pour analyser le fichier vectoriel.
CHEMIN_LINEAMENTS = DOSSIER_DONNEES / "lineaments.shp"
COUCHE_LINEAMENTS = None  # GeoPackage : None si une seule couche, sinon son nom exact.
PAS_ANGLE_ROSE = 5  # Largeur des classes d'orientation, en degrés.

# Réglages de préparation et d'entraînement.
TAILLE_PATCH = 64
PAS = 64
PROPORTIONS = (0.70, 0.15, 0.15)
GRAINE = 42
PERCENTILES = (2, 98)
TRANSFORMATIONS = ["original", "rotation90", "rotation180", "rotation270", "miroir_h", "miroir_v"]
EPOQUES = 150
BATCH_SIZE = 16
LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-4
DEVICE = "auto"  # "auto", "cpu" ou "cuda".
LIMITE_MEMOIRE_MIO = 2048  # Limite par opération, pas une garantie sur la RAM totale.

# Réglages d'affichage : aucune influence sur l'apprentissage.
NOMBRE_PATCHS_AFFICHES = 4  # Entre 1 et 32, limité aux patches disponibles.
CANAL_NORMALISATION = "MNT"
INDEX_PATCH = 0
ACTIVER_EXPLORATEUR = True

# Utilisés seulement en mode "modele_enregistre".
CHEMIN_MODELE = DOSSIER_SORTIE / "mon_entrainement/best.pt"
TEST_REFERENCE = False  # True : données exactes et partition sauvegardée par ce projet.
STATISTIQUES_HISTORIQUES = None  # Bornes par canal si absentes d'un ancien checkpoint.
CONFIANCE_CHECKPOINT = False  # True seulement pour vos propres anciens checkpoints de confiance.
```

Elle vérifie également la cohérence des réglages : mode reconnu, présence d'un masque pour l'entraînement, proportions valides et taille de patch compatible avec le U-Net.

## 4. Découper les rasters en patches

Les rasters complets sont généralement trop grands pour être fournis directement au U-Net. Le pipeline découpe donc toutes les couches et le masque en petites images appelées **patches**, en conservant exactement la même position dans chaque fichier.

Avec `TAILLE_PATCH=64` et `PAS=64`, chaque patch mesure 64 × 64 pixels et le suivant commence juste après le précédent. La surface réelle représentée dépend de la résolution : à 10 m par pixel, un patch couvre 640 × 640 m.

Cette cellule estime le nombre de patches, extrait les patches complets et affiche leur grille. Les contours verts désignent les patches contenant au moins un pixel de linéament ; les contours rouges désignent ceux qui ne contiennent que du fond. Les bordures trop petites sont écartées. Dans cette expérience, une valeur invalide dans une couche est signalée comme une erreur à corriger avant le découpage.

**À vérifier :** le nombre de patches doit être suffisant et la grille doit couvrir correctement la zone utile.


```python
if MODE == "entrainer":
    resume, figure = exp.apercu_decoupage()
    display(resume)
    plt.show()
    display(exp.decouper(limite_memoire_mio=LIMITE_MEMOIRE_MIO))
    figure = exp.visualiser_decoupage()
    plt.show()
else:
    print("Le modèle enregistré fournira son découpage à l'étape 9.")
```

## 6. Normaliser chaque canal

Les trois canaux n'ont pas la même unité : altitude, intensité d'ombrage et pente. Le pipeline calcule donc deux bornes séparées pour chaque canal à partir du train, puis normalise chaque canal entre 0 et 1. `CANAL_NORMALISATION` permet d'examiner successivement `MNT`, `Hillshade 315°` ou `Pente`.

Avec `PERCENTILES=(2, 98)`, le 2ᵉ percentile devient la borne basse et le 98ᵉ percentile la borne haute. Les valeurs sont ramenées dans `[0, 1]`. La validation et le test n'interviennent jamais dans le calcul de ces bornes.

La cellule affiche les bornes ainsi qu'une comparaison avant/après. **À vérifier :** les structures utiles doivent rester visibles et le canal normalisé ne doit pas devenir entièrement sombre, clair ou constant.

## 7. Augmenter les données d'entraînement

L'augmentation crée plusieurs orientations d'un même patch afin d'exposer le modèle à davantage de configurations géométriques. Avec le réglage proposé, chaque patch original produit six images : l'original, trois rotations et deux miroirs.

La transformation est appliquée simultanément à tous les canaux d'entrée et au masque. La position du linéament reste donc correcte après chaque rotation ou miroir.

Seul le train est augmenté. La validation et le test restent inchangés pour conserver une évaluation indépendante.

**Exemple :** 100 patches originaux deviennent 600 images d'entraînement avec six transformations. Le tableau et la figure permettent de vérifier ce comptage et la cohérence visuelle des transformations.
